# NYC Wellness Route Navigator

**Final Project - Data Mapping Systems**

This project creates personalized wellness routes through New York City by combining geospatial data analysis, social media sentiment analysis, and interactive mapping. The system analyzes NYC's cultural institutions, parks, and urban features to generate routes optimized for different personality types and wellness goals.

## Project Overview

The NYC Wellness Route Navigator addresses the challenge of finding meaningful, personalized routes through the city that promote mental wellness and cultural engagement. By analyzing geospatial data from NYC Open Data and incorporating social sentiment analysis, the system creates routes that match individual personality types and wellness preferences.

### Key Features:
- **Personality-Based Route Generation**: Routes tailored to Creative, Healing, Social, and Energetic personality types
- **Multi-Modal Data Integration**: Combines NYC geospatial data, cultural institutions, parks, and social media sentiment
- **Interactive Visualization**: Dynamic maps with route options and detailed place information
- **Social Context Analysis**: Incorporates community sentiment and trending topics for each location

## Data Sources and Methodology

### Primary Data Sources:
1. **NYC Open Data**: Cultural institutions, parks, building footprints
2. **NYC QGIS Data**: Geospatial layers for Manhattan
3. **Social Media Analysis**: Simulated sentiment analysis for location context
4. **Google Places API**: Additional venue information (optional)

### Technical Approach:
- **Geospatial Analysis**: Using GeoPandas and Folium for spatial operations
- **Route Optimization**: Distance-based algorithms for waypoint selection
- **Personality Scoring**: Custom algorithms that weight locations based on user preferences
- **Interactive Mapping**: Folium-based visualizations with custom markers and popups
